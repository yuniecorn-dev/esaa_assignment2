###행렬 분해의 이해

빈칸이 많은 평점 표를 수학적으로 어떻게 쪼개고 다시 채우는지

**1. 다항식의 인수분해와 같은 원리**

* 수학 시간에 복잡한 다항식 $x^2 + 5x + 6$을 두 개의 단순한 인수 $(x+2) * (x+3)$으로 인수분해 하듯이, 행렬 분해 역시 복잡한 다차원 매트릭스를 저차원 매트릭스로 쪼개는 작업입니다.



**2. 듬성듬성한 표(희소 행렬)를 꽉 찬 표(밀집 행렬)로 나누기**

* **$R$ 행렬:** $M \times N$ 차원으로 구성된 사용자-아이템 평점 행렬입니다. 아직 사용자들이 평가하지 않은 아이템이 많아 널(NaN) 값을 많이 가지는 고차원의 '희소 행렬'입니다.


* 이 $R$ 행렬을 사용자와 잠재 요인의 관계를 나타내는 **$P$ 행렬**, 그리고 아이템과 잠재 요인의 관계를 나타내는 **$Q$ 행렬**(계산을 위해 전치 행렬인 $Q.T$로 변환)로 분해합니다. 이렇게 분해된 $P$와 $Q$ 행렬은 저차원의 '밀집 행렬'이 됩니다.



**3. 분해된 행렬을 곱해 안 본 영화의 평점 예측하기**

* 행렬 분해를 통해 얻어진 $P$ 행렬과 $Q.T$ 행렬을 내적(곱하기)하면, 미정 값(NaN)을 포함한 모든 평점 값이 예측 평점으로 다시 계산된 새로운 매트릭스($\hat{R}$)를 얻을 수 있습니다.


* 예를 들어, User 2가 아직 평점을 매기지 않은 Item 3의 점수를 예측해 볼 수 있습니다. $P$ 행렬에 있는 User 2의 벡터와 $Q$ 행렬에 있는 Item 3의 벡터를 곱하고 더하면($2.14 \times 1.41 + 0.08 \times 0.14$) **3.02점**이라는 예측값이 나옵니다.



**4. 빈칸이 많은데 어떻게 분해할까? (SVD의 한계 극복)**

* 행렬 분해에는 주로 SVD(Singular Value Decomposition) 방식을 이용합니다.


* 하지만 일반적인 SVD는 널(NaN) 값이 없는 행렬에만 적용할 수 있다는 단점이 있습니다.


* 실제 평점 행렬($R$)에는 아직 평점이 매겨지지 않은 빈칸이 매우 많기 때문에, 확률적 경사 하강법(SGD)이나 ALS 방식을 이용하여 SVD를 수행합니다.



요약하자면, **빈칸 투성이인 원래의 평점표($R$)를 SGD 같은 방식을 써서 사용자 특성표($P$)와 아이템 특성표($Q$)로 분해한 뒤, 이 둘을 다시 곱해서 모든 빈칸의 예상 평점이 채워진 추천 리스트($\hat{R}$)를 만들어내는 수학적 과정**

###확률적 경사 하강법을 이용한 행렬 분해

#### 확률적 경사 하강법(SGD)을 이용한 최적화
* 평점이 없는 널(NaN) 값이 많은 행렬은 일반적인 SVD 방식으로 분해하기 어려우므로 SGD 방식을 주로 사용함.
* 컴퓨터가 정답(최적의 행렬)을 찾아가기 위해 다음 4단계를 반복함.
  1. $P$와 $Q$ 행렬에 임의의 랜덤 값을 설정함.
  2. 예측 평점을 계산하고 실제 평점과의 오류(차이) 값을 계산함.
  3. 이 오류를 줄이는 방향으로 $P$와 $Q$ 값을 미세하게 수정(업데이트)함.
  4. 오차가 만족할 만큼 줄어들 때까지 위 작업을 무한 반복함.

---
#### 비용 함수 (Cost Function)
* 모델이 최적의 $P$와 $Q$를 찾기 위해 지속적으로 최소화해야 하는 목표 공식을 의미함.

$$ \min \sum (r_{u,i} - p_u q_i^t)^2 + \lambda(\|q_i\|^2 + \|p_u\|^2) $$

* $(r_{u,i} - p_u q_i^t)^2$: 실제 평점과 예측 평점 간의 차이(오류)를 최소화함.
* $\lambda(\|q_i\|^2 + \|p_u\|^2)$: 학습 데이터에만 억지로 맞춰지는 과적합 현상을 피하기 위해 L2 규제(Regularization)를 적용함.
* 여기서 $\lambda$(람다)는 규제의 강도를 결정하는 계수 역할을 수행함.

---

####가중치 업데이트 수식 (Update Rule)
* 비용 함수를 최소화하기 위해 $P$와 $Q$의 값을 구체적으로 어떻게 변경할지 결정하는 수식을 뜻함.

$$ e_{u,i} = r_{u,i} - \hat{r}_{u,i} $$
$$ p_u' = p_u + \eta(e_{u,i} * q_i - \lambda * p_u) $$
$$ q_i' = q_i + \eta(e_{u,i} * p_u - \lambda * q_i) $$

* $e_{u,i}$: 실제 평점과 예측 평점의 차이인 오류 값을 의미함.
* $\eta$(에타): SGD 학습률(Learning Rate)로, 한 번에 값을 얼마나 크게 이동시킬지(수정할지) 결정함.
* 계산된 오류 값($e$)과 학습률($\eta$)을 반영해 기존의 $p_u$와 $q_i$ 값을 정답에 가깝도록 계속해서 업데이트함.

```python
import numpy as np

# 원본 행렬 R 생성, 분해 행렬 P와 Q 초기화, 잠재 요인 차원 K는 3으로 설정.
R = np.array([[4, np.nan, np.nan, 2, np.nan],
              [np.nan, 5, np.nan, 3, 1],
              [np.nan, np.nan, 3, 4, 4],
              [5, 2, 1, 2, np.nan]])
num_users, num_items = R.shape
K=3

# P와 Q 행렬의 크기를 지정하고 정규 분포를 가진 임의의 값으로 입력합니다.
np.random.seed(1)
P = np.random.normal(scale=1./K, size=(num_users, K))
Q = np.random.normal(scale=1./K, size=(num_items, K))
```

```python
from sklearn.metrics import mean_squared_error

def get_rmse(R, P, Q, non_zeros):
  error=0
  # 두 개의 분해된 행렬 P와 Q.T의 내적으로 예측 R 행렬 생성
  full_pred_matrix = np.dot(P, Q.T)

  # 실제 R 행렬에서 Null이 아닌 값의 위치 인덱스 추출해 실제 R 행렬과 예측 행렬의 RMSE 추출
  x_non_zero_ind = [non_zero[0] for non_zero in non_zeros]
  y_non_zero_ind = [non_zero[1] for non_zero in non_zeros]
  R_non_zeros = R[x_non_zero_ind, y_non_zero_ind]
  full_pred_matrix_non_zeros = full_pred_matrix[x_non_zero_ind, y_non_zero_ind]
  mse = mean_squared_error(R_non_zeros, full_pred_matrix_non_zeros)
  rmse = np.sqrt(mse)

  return rmse
```


* SGD(확률적 경사 하강법)를 기반으로 행렬 분해를 수행함.


* 가장 먼저 R 행렬에서 널(Null) 값을 제외한 데이터의 행렬 인덱스를 추출함.


* 주요 파라미터로 `steps`(반복 업데이트 횟수), `learning_rate`(학습률), `r_lambda`(L2 정규화 계수)를 설정함.


* `steps=1000`번 반복하며, $p'_u = p_u + \eta(e_{(u,i)} * q_i - \lambda * p_u)$ 및 $q'_i = q_i + \eta(e_{(u,i)} * p_u - \lambda * q_i)$ 수식을 통해 새로운 $p_u$, $q_i$ 값으로 업데이트함.


* `get_rmse()` 함수를 사용하여 50회 반복할 때마다 계산된 오류 값을 출력함.

```python
# R > 0인 행 위치, 열 위치, 값을 non_zeros 리스트에 저정.
non_zeros = [ (i, j, R[i, j]) for i in range(num_users) for j in range(num_items) if R[i, j] > 0 ]

steps=1000
learning_rate=0.01
r_lambda=0.01

# SGD 기법으로 P와 Q 매트릭스를 계속 업데이트.
for step in range(steps):
  for i, j, r in non_zeros:
    # 실제 값과 예측 값의 차이인 오류 값 구함
    eij = r - np.dot(P[i, :], Q[j, :].T)
    # Regularization을 반영한 SGD 업데이트 공식 적용
    P[i, :] = P[i, :] + learning_rate*(eij * Q[j, :] - r_lambda*P[i, :])
    Q[j, :] = Q[j, :] + learning_rate*(eij * P[i, :] - r_lambda*Q[j, :])

  rmse = get_rmse(R, P, Q, non_zeros)
  if (step % 50) == 0:
    print("### iteration step : ", step, "rmse : ", rmse)
```

```python
pred_matrix = np.dot(P, Q.T)
print('예측 행렬:\n', np.round(pred_matrix, 3))
```
